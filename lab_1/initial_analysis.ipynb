## Importing packages + reading in the data

```python
# 1. Load the Foreign Gifts data.
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('ForeignGifts_edu.csv', low_memory=False)
df.head()
```

2. Describe the differences between different classes of gift types.

Contracts are the most common type and contribute the most money overall. Monetary gifts are less frequent and have a lower mean and median. Real-estate gifts are extremely rare (11 records), but their mean and median are the largest, so their typical recorded value is higher.

3. Answer the following questions:

<div class="alert alert-block alert-warning">
<b>Question 1:</b> Which country gives the most money, in total?
</div>

```python
# Add the gift amounts for each country.
country_totals = df.groupby('Country of Giftor')['Foreign Gift Amount'].sum()
country_totals = country_totals.sort_values(ascending=False)
country_totals.head(10)
```

**Answer:** Qatar gives the most money in total, with **2,706,240,869 dollars**.

<div class="alert alert-block alert-warning">
<b>Question 2:</b> Which country initiates the most gifts, by count?
</div>

```python
# Count how many rows belong to each country.
country_counts = df['Country of Giftor'].value_counts()
country_counts.head(10)
```

**Answer:** England initiates the most gifts, with **3,655** records.

<div class="alert alert-block alert-warning">
<b>Question 3:</b> Which country gives the largest gifts, on average?
</div>

```python
# Find the average gift for each country.
country_average = df.groupby('Country of Giftor')['Foreign Gift Amount'].mean()
country_average = country_average.sort_values(ascending=False)
country_average.head(10)
```

**Answer:** Bermuda has the largest average gift, approximately **7.69 million dollars**. Counts are shown because averages for countries with few gifts can be unstable.

## Ordering by institution

<div class="alert alert-block alert-warning">
<b>Question 4:</b> Which institution receive the most money, in total?
</div>

<div class="alert alert-block alert-warning">
<b>Question 5:</b> Which institution receive the greatest number of gifts, by count?
</div>

```python
# Add the gift amounts for each institution.
institution_totals = df.groupby('Institution Name')['Foreign Gift Amount'].sum()
institution_totals = institution_totals.sort_values(ascending=False)

# Count the number of gifts for each institution.
institution_counts = df['Institution Name'].value_counts()

print(institution_totals.head(1))
print(institution_counts.head(1))
```

```python
top_20_institutions = institution_totals.head(20).sort_values()

plt.figure(figsize=(10, 8))
sns.barplot(
    x=top_20_institutions.values,
    y=top_20_institutions.index,
    color='steelblue'
)
plt.title('Top 20 Institutions by Total Foreign Gift Amount')
plt.xlabel('Total foreign gift amount ($)')
plt.ylabel('Institution')
plt.tight_layout()
plt.show()
```

**Answers:** Carnegie Mellon University receives the most money (**1,477,922,504 dollars**). The University of California, Los Angeles receives the most gifts (**3,916**).



<div class="alert alert-block alert-warning">
<b>Question 6:</b> What is the average amount each university receives? Median amount? If they are different, why?
</div>

```python
# Use the total received by each university.
amount_per_university = institution_totals
mean_received = amount_per_university.mean()
median_received = amount_per_university.median()
print(f'Mean received per university:   ${mean_received:,.2f}')
print(f'Median received per university: ${median_received:,.2f}')
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(amount_per_university, bins=30, kde=True, color='darkgreen')
plt.axvline(mean_received, color='red', linestyle='--', linewidth=2, label=f'Mean: ${mean_received:,.0f}')
plt.axvline(median_received, color='black', linestyle='-', linewidth=2, label=f'Median: ${median_received:,.0f}')
plt.title('Distribution of Total Foreign Giving by Institution')
plt.xlabel('Total foreign gift amount per institution ($)')
plt.ylabel('Number of institutions')
plt.legend()
plt.tight_layout()
plt.show()
```

**Answer:** The mean is **52,202,878.97 dollars** per university and the median is **6,486,791.50 dollars**. The distribution is strongly skewed. Some universities receive extremely large amounts, bringing the mean much higher, while the median is less affected by outliers.


<div class="alert alert-block alert-warning">
<b>Question 7:</b> What is the largest flow of a single country to a single institution?
</div>



```python
# Make a table of money sent from each country to each institution.
country_institution = pd.crosstab(
    df['Country of Giftor'],
    df['Institution Name'],
    values=df['Foreign Gift Amount'],
    aggfunc='sum'
)
country_institution.head()
```

```python
# Turn the table into one long list and sort it.
flows = country_institution.stack().sort_values(ascending=False)
flows.head(10)
```

**Answer:** The largest country-to-institution flow is **Qatar to Cornell University**, totaling **1,018,473,315 dollars**.

<div class="alert alert-block alert-warning">
<b>Question 8:</b> Which are the top giftors to US academic institutions?
</div>

```python
# Remove rows where the giftor name is missing.
known_giftors = df.dropna(subset=['Giftor Name'])

# Find the giftors who gave the most money.
giftor_totals = known_giftors.groupby('Giftor Name')['Foreign Gift Amount'].sum()
giftor_totals = giftor_totals.sort_values(ascending=False)
print('Top giftors by total amount:')
print(giftor_totals.head(10))

# Count how many gifts came from each giftor.
giftor_counts = known_giftors['Giftor Name'].value_counts()
print('\nTop giftors by number of gifts:')
print(giftor_counts.head(10))
```

**Answer:** By total amount, the leading named giftor is **Qatar Foundation** (**1,166,503,744 dollars**). By count, it is **Saudi Arabian Cultural Mission** (**480 gifts**). All recipients in this education dataset are U.S. institutions, so no recipient-country filter is needed.

```python
import plotly.graph_objects as go

giftor = 'Giftor Name'
#giftor = 'Country of Giftor'
recipi = 'Institution Name'
flow = 'Foreign Gift Amount'
N = 25

flows = (
    df.groupby([giftor, 
                recipi])
      [flow]
      .sum()
      .nlargest(N)
      .reset_index()
)

labels = (
    flows[giftor].tolist()
    + flows[recipi].tolist()
)

labels = list(dict.fromkeys(labels))

fig = go.Figure(
    go.Sankey(
        node=dict(label=labels),
        link=dict(
            source=flows[giftor]
                        .map(labels.index),
            target=flows[recipi]
                        .map(labels.index),
            value=flows[flow]
        )
    )
)

fig.show()
```
